```python
# make jobs.txt
problems   = [1, 2, 3]
sizes      = [30, 60, 125, 250, 500]
models     = ['bn']
optimizers = ['cmaes', 'turbo1', 'botorch', 'baxus', 'de', 'one_plus_one']

with open('jobs_bn.txt', 'w') as f:
    for problem in problems:
        for size in sizes:
            for model in models:
                for optimizer in optimizers:
                    f.write(f"{problem} {size} {model} {optimizer}\n")

```

```python
# make jobs with 15 iterations\
import itertools

problems = [1,2,3]  
sizes = [30,60,125,250,500]             
models = ["rsm", "rf", "xgb", "gp"]#,"bn"]
# models = ["bn"]
optimizers = ["one_plus_one"]               
SEEDS = list(range(1312, 1312+15))                   


file_p_1and2 = "jobs_p_1and2_opo.txt"
file_p3  = "jobs_long_opo.txt"

p_1and2 = []
p3  = []

for seed in SEEDS:
    for problem in problems:
        for size in sizes:
            for model in models:
                for optimizer in optimizers:
                    if problem==3:
                        p3.append(f"{problem} {size} {model} {optimizer} {seed}\n")
                    else:
                        p_1and2.append(f"{problem} {size} {model} {optimizer} {seed}\n")


with open(file_p_1and2, "w") as f:
    f.writelines(p_1and2)

with open(file_p3, "w") as f:
    f.writelines(p3)


```

```python
# make mechbench optimizer runs jobs
SEEDS = list(range(1312, 1312+15))
problems = [1,2,3]
optimizers = ['de', 'one_plus_one', 'cmaes', 'botorch', 'baxus', 'turbo1']
jobs = []
for seed in SEEDS:
    for problem in problems:
        for optimizer in optimizers:
            jobs.append(f"{problem} {optimizer} {seed}\n")

with open('mechbench_optimizer_runs_jobs.txt', 'w') as f:
    f.writelines(jobs)
```

```python
missing_first = []
missing_second = []

with open("missing_jobs_mechbench.txt", "r") as missing_all:
    jobs = [line.strip() for line in missing_all.readlines()]
    missing_first = [job.strip() for job in jobs[:len(jobs)//2]]
    missing_second = [job.strip() for job in jobs[len(jobs)//2:]]


with open("missing_foppe.txt", "w") as first_file:
    for job in missing_first:
        first_file.write(f'{job}\n')

with open("missing_odysseas.txt", "w") as second_file:
    for job in missing_second:
        second_file.write(f'{job}\n')
```
